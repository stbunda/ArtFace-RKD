# Data Preperation
In this file there are several snippets that can be used to create relevant lists/dictionaries and the folder 'sitters' that contains the faces organized by sitter_id

```python
import pandas as pd
import chardet

with open("../data2/HP_labels_v8_source.csv", "rb") as f:
    result = chardet.detect(f.read(100000))

print(result)
archive = pd.read_csv('../data2/HP_labels_v8_source.csv', encoding=result['encoding'], sep=";")
archive.head()
```

```python
from pathlib import Path

folder_path = Path("../data2/historical_faces/")
painting_files = list(folder_path.glob("*.jp*g"))

faces_files = {file.stem.split('-')[0]: file.name for file in painting_files}

print(faces_files)
len(faces_files)
```

```python
from pprint import pprint

########### List with Sitters ###########
"""
This snippet creates a dictionary with the sitter_id (key) and a list of available prirefs containing this sitter (value)
    zitters
"""
zitters = {}
for index, painting in archive.iterrows():
    filename = painting['filename'].rsplit('.', 1)[0]
    zitter = painting['pid']
    if zitter not in zitters:
        zitters[zitter] = []
    zitters[zitter].append(faces_files[filename])
pprint(zitters)
```

```python
########### Sitters with more than one painting ###########
"""
This snippet lists all the sitters that show up on more than one painting
"""
counter = 0
counter2 = 0
dicter = []
for person, prirefs in zitters.items():
    if len(prirefs) > 1:
        print(f"{counter}  --  Person {person} has {len(prirefs)} paintings: {prirefs}")
        counter2 += len(prirefs)
        counter += 1
        for img in prirefs:
            dicter.append(img)

print(counter2)
```

```python
for img in list(faces_files.values()):
    if img not in dicter:
        print(img)
```

```python
########### Personen directory ###########
"""
This snippet creates a new directory containing all the faces organized by sitter
"""

import shutil
from pathlib import Path

already_processed = []

for persoon, afbeeldingen in zitters.items():
    print(persoon, afbeeldingen)
    for img in afbeeldingen:
        print(f'\t{img}')
        source_path = Path(f"../data2/historical_faces/{img}")
        destination_path = Path(f"../data2/historical_sitters/{persoon}/")
        destination_path.mkdir(parents=True, exist_ok=True)
        shutil.copy(source_path, destination_path / f'{source_path.name}')
        # if img == []:
        #     continue
        # gezichten = afbeeldingen_gezichten[img[0]]
        # for gezichten_nr in gezichten:
        #     source_path = Path(f'../data/faces/{gezichten_nr}.jpg')
        #     destination_folder = Path(f'../data/sitters/{persoon}')
        #
        #     destination_folder.mkdir(parents=True, exist_ok=True)
        #     shutil.copy(source_path, destination_folder / f'{source_path.name}')
```
