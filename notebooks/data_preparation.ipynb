# Data Preperation
In this file there are several snippets that can be used to create relevant lists/dictionaries and the folder 'sitters' that contains the faces organized by sitter_id

```python
import xml.etree.ElementTree as ET

# The data is structured in an XML-tree
tree = ET.parse('../data/rkd_metadata.xml')
root = tree.getroot()
```

```python
from pathlib import Path

folder_path = Path("../data/paintings/")
painting_files = list(folder_path.glob("*.jp*g"))

paintings = {file.stem: file.name for file in painting_files}

print(paintings)
```

```python
########### List with available Paintings ###########
"""
This snippet creates a dictionary with the available priref (key) and the name of the jpg.stem (value)
Available prirefs are the prirefs with paintings in data/paintings
    prirefs_beschikbaar
"""

prirefs = {}
for record in root:
    priref = record.findtext('priref')
    prirefs[priref] = []
    for media in record.findall('.//original_file_name'):
        file_name = media.text
        if file_name in paintings:
            prirefs[priref].append(file_name)

prirefs_beschikbaar = {k: v for k, v in prirefs.items() if v != []}
print(len(prirefs_beschikbaar), prirefs_beschikbaar)
```

```python
########### List with Sitters ###########
"""
This snippet creates a dictionary with the sitter_id (key) and a list of available prirefs containing this sitter (value)
    zitters
"""
zitters = {}
for record in root:
    priref = record.findtext('priref')
    for media in record.findall('.//persoonsnummer'):
        if media.text != '\n':
            persoonsnummer = media.text
            if persoonsnummer not in zitters:
                zitters[persoonsnummer] = []
            zitters[persoonsnummer].append(prirefs[priref])

print(zitters)
```

```python
########### Sitters with more than one painting ###########
"""
This snippet lists all the sitters that show up on more than one painting
"""
counter = 0
for person, prirefs in zitters.items():
    if len(prirefs) > 1:
        counter += 1
        print(f"{counter}  --  Person {person} has {len(prirefs)} paintings: {prirefs}")
```

```python
########### Paintings with faces ###########
"""
This snippet lists all the sitters in each painting
    afbeeldingen_gezichten
"""
from pathlib import Path

folder_path = Path("../data/faces/")
faces = list(folder_path.glob("*.jp*g"))
afbeeldingen_gezichten = {file.stem: [] for file in painting_files}


face_files = [jpg.stem for jpg in faces]

for gezicht in face_files:
    afbeelding = gezicht.split('-')[0]
    afbeeldingen_gezichten[afbeelding].append(gezicht)
```

```python
########### Personen directory ###########
"""
This snippet creates a new directory containing all the faces organized by sitter
"""

import shutil
from pathlib import Path

already_processed = []

for persoon, afbeelding in zitters.items():
    print(persoon, afbeelding)
    for img in afbeelding:
        if img == []:
            continue
        gezichten = afbeeldingen_gezichten[img[0]]
        for gezichten_nr in gezichten:
            source_path = Path(f'../data/faces/{gezichten_nr}.jpg')
            destination_folder = Path(f'../data/sitters/{persoon}')

            destination_folder.mkdir(parents=True, exist_ok=True)
            shutil.copy(source_path, destination_folder / f'{source_path.name}')
```

```python
########### Meerdere Gezichten Personen directory ###########
"""
This snippet creates a new directory containing all the faces organized by sitter
"""

import os
import shutil

def copy_folders_with_min_images(src_dir, dst_dir, min_images=2, image_exts={'.jpg', '.jpeg', '.png', '.bmp', '.gif'}):
    os.makedirs(dst_dir, exist_ok=True)

    for folder_name in os.listdir(src_dir):
        folder_path = os.path.join(src_dir, folder_name)

        if os.path.isdir(folder_path):
            # Count image files in this folder
            image_files = [f for f in os.listdir(folder_path)
                           if os.path.splitext(f)[1].lower() in image_exts]

            if len(image_files) >= min_images:
                dst_folder_path = os.path.join(dst_dir, folder_name)
                shutil.copytree(folder_path, dst_folder_path)
                print(f"Copied {folder_name} ({len(image_files)} images)")
            else:
                print(f"Skipped {folder_name} ({len(image_files)} images)")

# Example usage
src_folder = "../data/sitters"
dst_folder = "../data/sitters_multi_face"
copy_folders_with_min_images(src_folder, dst_folder, min_images=2)
```
